# Getting Started with coursekata

Welcome! This interactive notebook demonstrates the key features of the

Run each cell by clicking it and pressing **Shift+Enter**.

## 1. Loading the Package

The `coursekata` package bundles everything you need for introductory statistics.
Loading it gives you access to data, plotting, and modeling functions.

```r
library(coursekata)
```

## 2. Exploring a Dataset

The `penguins` dataset contains measurements for penguin species from the
Palmer Archipelago in Antarctica. Let's take a look at the first few rows
and the structure of the data.

```r
head(penguins)
```

```r
str(penguins)
```

## 3. Creating a Visualization

The `gf_` family of functions makes it easy to create plots using a formula
interface. Here we create a histogram of bill lengths and a scatter plot
colored by species.

```r
gf_histogram(~bill_length_cm, data = penguins, bins = 20)
```

```r
gf_point(body_mass_kg ~ bill_length_cm, data = penguins, color = ~species)
```

## 4. Fitting a Model

Use `lm()` to fit a linear model and `supernova()` to get a clean ANOVA
table. Here we model body mass as a function of bill length.

```r
model <- lm(body_mass_kg ~ bill_length_cm, data = penguins)
model
```

```r
supernova(model)
```
